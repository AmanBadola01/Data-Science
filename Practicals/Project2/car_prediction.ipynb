## Issues in data
1. Year has many non-year values
2. Year object to int
3. Price has Ask for Price
4. Kms_drivern has kms with integers
5. kms_driven object to int
6. kms_driven has nan(Not a Number) values
7. fuel_type has nan values
8. Inconsistent Names -> converting into catergorical (Keeping first 3 words of name)

```python
## Cleaning the data 
''' 1 & 2: Removing year's non year values and changing year type to int '''
backup = car.copy()
car= car[car['year'].str.isnumeric()] #Removed the non year values 
car['year'] = car['year'].astype(int) #changed the datatype to int
```

```python
''' 3. Removing "Ask for Price" in price column'''
car = car[car['Price']!= "Ask For Price"]
```

```python
car["Price"] = car['Price'].str.replace(",", "")   # Replaced "," -> ""
```

```python
car['Price']
```

```python
car["Price"] = car['Price'].astype(int)        #Convering object to int data type
```

```python
''' 4. Remove kms from Kms_driven'''
car['kms_driven'] = car['kms_driven'].str.replace('kms', '')    #Removed kms
```

```python
car['kms_driven'] = car['kms_driven'].str.replace(',','')    #Replaced "," -> ""
```

```python
car = car[car['kms_driven']!= "Petrol"] #Removed Petrol from kms_driven
```

```python
car["kms_driven"] = car['kms_driven'].astype(int) #Change data type from object to int
```

```python
car.info()
```

```python
car['kms_driven']
```

```python
car= car[~car["fuel_type"].isna()] #Removing row which has nan value in it
```

```python
''' Altering Names '''
car['name'].str.split(' ').str.slice(0,3).str.join(' ') #Taking only first 3 names from name column
```

```python
car= car.reset_index(drop=True)
```

```python
#Finding outliar
car= car[car['Price'] <6e6].reset_index(drop=True)
```

##### MODEL BUILDING


```python
car.to_csv("cleaned_car.csv")
```

```python
x = car.drop(columns="Price")
y = car['Price'] 
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test= train_test_split(x,y, train_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_transformer
from sklearn.pipeline import make_pipeline
```

```python
ohe = OneHotEncoder()
ohe.fit(x[['name', 'company', 'fuel_type']])
```

```python
column_trans = make_column_transformer((OneHotEncoder(categories = ohe.categories_), ['name','company','fuel_type']), 
                                       remainder='passthrough')
```

```python
lr = LinearRegression()
```

```python
pipe= make_pipeline(column_trans, lr)
```

```python
pipe.fit(x_train, y_train)
```

```python
y_pred = pipe.predict(x_test)
```

```python
r2_score(y_test, y_pred)
```

```python
import numpy as np

scores = []
for i in range(1000):
    x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=i)
    lr = LinearRegression()
    pipe = make_pipeline(column_trans, lr)
    pipe.fit(x_train, y_train)
    y_pred = pipe.predict(x_test)
    scores.append(r2_score(y_test, y_pred))
    
```

```python
np.argmax(scores)
```

```python
scores[np.argmax(scores)]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2, random_state= np.argmax(scores))
lr = LinearRegression()
pipe = make_pipeline(column_trans, lr)
pipe.fit(x_train, y_train)
y_pred = pipe.predict(x_test)
r2_score(y_test, y_pred)
```
